## Distribution of Patient Counts Across Categorical Features: Bar and Pie Charts

```python
# Seperated Object Columns
object_columns = df.select_dtypes(include = object)
object_columns.columns.tolist()
```

```python
cols = ['Gender','Blood Type', 'Medical Condition','Insurance Provider', 'Admission Type',
        'Medication', 'Test Results']
```

```python
sns.set_theme(style="ticks")

for i in cols:
    fig, ax = plt.subplots(1, 2, figsize=(12, 5))
    fig.suptitle(f'** {i} **', fontsize=20)

    # Bar chart
    df[i].value_counts().plot(
        kind='bar',
        color=sns.color_palette("tab10"),
        ax=ax[0]
    )
    ax[0].set_title("Bar Plot")
    ax[0].set_ylabel('Number of Patients')
    
    # Pie chart
    df[i].value_counts().plot(
        kind='pie',
        autopct="%.2f%%",
        ax=ax[1]
    )
      
    ax[1].set_title("Pie Chart")

    plt.tight_layout()
    plt.show()  
```

## Identifying Key Categories Impacting Billing Revenue

```python
cols = ['Gender','Blood Type', 'Medical Condition','Doctor',
        'Hospital', 'Insurance Provider', 'Admission Type',
        'Medication', 'Test Results']
```

```python
# Initialize Plotly notebook mode to display plots inline in Jupyter Notebooks
pyo.init_notebook_mode(connected=True)

for i in cols:
    
    if i == 'Gender':
        # Group by 'Gender' and calculate the sum of 'Billing Amount'
        char_bar = df.groupby(['Gender'])[['Billing Amount']].sum().reset_index()
    else:
        # Group by the current column (i) and calculate the sum of 'Billing Amount'
        char_bar = df.groupby([i])[['Billing Amount']].sum().reset_index()

     # Sort the dataframe by 'Billing Amount' in descending order to get the top values
    char_bar = char_bar.sort_values(by="Billing Amount", ascending=False)
    top = char_bar.head(10)

    # Create Bar chart
    fig = go.Figure()

    fig.add_trace(go.Bar(
        x=top[i if i != 'Gender' else 'Gender'],
        y=top['Billing Amount'],
        
    ))
    
    # Customize the layout of the chart
    fig.update_layout(
        title=f"{i} by Billing Amount", 
        xaxis_title=i if i != 'Gender' else 'Gender',
        yaxis_title="Billing Amount",
        plot_bgcolor='white',
        paper_bgcolor='white',
        font=dict(color='black'),
        bargap=0.3
    )


    pyo.iplot(fig)
```

## Billing Amount Distribution by Medical Condition and Prescribed Medication

```python
df_trans = df.groupby(['Medical Condition', 'Medication'])[['Billing Amount']].sum().reset_index()

plt.figure(figsize=(15,6))
sns.barplot(x=df_trans['Medical Condition'], y=df_trans['Billing Amount'], hue=df_trans['Medication'], ci=None, palette="Set1")
plt.title("Billing Amount according to Medical Condition and Medication")
plt.ylabel("Billing Amount")
plt.xticks(rotation=45, fontsize = 9)
plt.show()
```

## Billing Amount Distribution Based on Medical Condition and Test Results

```python
df_trans = df.groupby(['Medical Condition', 'Test Results'])[['Billing Amount']].sum().reset_index()

plt.figure(figsize=(15,6))
sns.barplot(x=df_trans['Medical Condition'], y=df_trans['Billing Amount'], hue=df_trans['Test Results'], ci=None, palette="Set1")
plt.title("Billing Amount according to Medical Condition and Test Results")
plt.ylabel("Billing Amount")
plt.xticks(rotation=45, fontsize = 9)
plt.show()
```

```python
# Creating a Column for Number of Days Hospitalized

df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'])

df['Days hospitalized'] = (df['Discharge Date'] - df['Date of Admission']).dt.days
df.head()
```

## Top Features by Average Number of Days Hospitalized

```python
# Initialize plotly offline mode
pyo.init_notebook_mode(connected=True)

for i in cols:
    
    if i == 'Gender':
        # Group by Gender and calculate mean of Days hospitalized
        char_bar = df.groupby(['Gender'])[['Days hospitalized']].mean().reset_index()
    else:
        # Group by other columns and calculate mean of Days hospitalized
        char_bar = df.groupby([i])[['Days hospitalized']].mean().reset_index()

    # Sort by average Days hospitalized in descending order
    char_bar = char_bar.sort_values(by="Days hospitalized", ascending=False)
    top = char_bar.head(10)

    # Create Bar chart
    fig = go.Figure()

    fig.add_trace(go.Bar(
        x=top[i if i != 'Gender' else 'Gender'],
        y=top['Days hospitalized']
    ))

    # Update layout
    fig.update_layout(
        title=f"Highest {i} according to average number of days hospitalized",
        xaxis_title=i if i != 'Gender' else 'Gender',
        yaxis_title="Days hospitalized",
        plot_bgcolor='white',
        paper_bgcolor='white',
        font=dict(color='black')
    )

    # Show figure
    pyo.iplot(fig)
```
